```python
import os
import glob
import sys
import tifffile
from tqdm import tqdm

# --- CONFIGURATION: SET YOUR FOLDER PATH HERE ---
folder_path = r"/home/students/rawat/czi_output/processed_images/"

# Validate folder exists
if not os.path.isdir(folder_path):
    print(f"❌ Error: Folder '{folder_path}' not found.")
    sys.exit(1)

# --- FIND ALL TIF FILES ---
tif_files = sorted(glob.glob(os.path.join(folder_path, "*.tif"))) + \
            sorted(glob.glob(os.path.join(folder_path, "*.ome.tif")))

if not tif_files:
    print(f"⚠️  No TIF files found in '{folder_path}'")
    sys.exit(0)

print(f"Found {len(tif_files)} TIF files in '{folder_path}'\n")
print("=" * 100)
print(f"{'Filename':<60} {'File Size (MB)':<20} {'Array Shape':<20}")
print("=" * 100)

total_size_mb = 0

for tif_path in tqdm(tif_files, desc="Processing"):
    filename = os.path.basename(tif_path)
    
    # File size on disk
    file_size_bytes = os.path.getsize(tif_path)
    file_size_mb = file_size_bytes / (1024 ** 2)
    total_size_mb += file_size_mb
    
    # Array shape in memory
    try:
        img = tifffile.imread(tif_path)
        array_shape = img.shape
        print(f"{filename:<60} {file_size_mb:<20.2f} {str(array_shape):<20}")
    except Exception as e:
        print(f"{filename:<60} {file_size_mb:<20.2f} {'ERROR: ' + str(e)[:15]:<20}")

print("=" * 100)
print(f"Total disk size: {total_size_mb:.2f} MB")
```
