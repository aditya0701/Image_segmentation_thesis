```python
import tifffile, numpy as np, os

path = r"D:\Thesis_images\Inference_Test_Outputs\MicroSAM_L_Original\Brain1_1340_Cont_AiryscanSR_Fly1_CalyxL_1100_masks.tif"
man_path = r"D:\Thesis_images\Final_dataset\all\test\masks\Brain1_1340_Cont_AiryscanSR_Fly1_CalyxL_1100_mask.tif"

with tifffile.TiffFile(path) as t:
    page = t.pages[0]
    arr  = t.asarray()
    print("=== INFERRED ===")
    print(f"dtype       : {arr.dtype}")
    print(f"shape       : {arr.shape}")
    print(f"unique vals : {np.unique(arr)}")
    print(f"compression : {page.compression}")
    print(f"bits/sample : {page.bitspersample}")
    print(f"file size   : {os.path.getsize(path)/1e6:.2f} MB")
    print(f"raw size    : {arr.nbytes/1e6:.2f} MB  (uncompressed)")
    has_ome = t.is_ome
    print(f"is OME-TIFF : {has_ome}")
    if has_ome:
        xml = t.ome_metadata
        print(f"OME XML size: {len(xml)/1e3:.1f} KB")

print()
with tifffile.TiffFile(man_path) as t:
    page = t.pages[0]
    arr  = t.asarray()
    print("=== MANUAL ===")
    print(f"dtype       : {arr.dtype}")
    print(f"shape       : {arr.shape}")
    print(f"compression : {page.compression}")
    print(f"bits/sample : {page.bitspersample}")
    print(f"file size   : {os.path.getsize(man_path)/1e6:.2f} MB")
    print(f"raw size    : {arr.nbytes/1e6:.2f} MB  (uncompressed)")
```
